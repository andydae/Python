```python
"""
IF, ELSEIF, ELSE: Python's version of CASE WHEN sql statements
"""

score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(grade)  # B

# Indentation matters here. The indented lines (4 spaces) are what belongs to each branch.
# There’s no BEGIN/END or {}. The colon : at the end of the line starts the block.
```

```python
"""
Loops
"""

# for loops
# A for loop runs once per item, like processing each row of a result set:

lures = ["Spinner", "Jig", "Crankbait"]

for lure in lures:
    print(f"In stock: {lure}")
    

# range() loops a set number of times:
for i in range(5):     # 0, 1, 2, 3, 4
    print(i)

for i in range(1, 6):  # 1 through 5 (end is excluded)
    print(i)
```

```python
# Loop plus if is like SELECT ... WHERE:

prices = [12.99, 4.50, 25.00, 8.75]

for p in prices:
    if p > 10:
        print(f"${p:.2f} is premium")
```

```python
# while loops
# Repeats as long as a condition is true. Less common in data work, but useful to know:

stock = 3
while stock > 0:
    print(f"Selling one, {stock} left")
    stock = stock - 1   # or: stock -= 1

# If the condition never becomes false, it loops forever. Stop it with Ctrl + C.
```

```python
# Exercise
# Given this list of order amounts:
orders = [45.00, 120.50, 8.99, 250.00, 63.25]

# Loop through it and print each order with a label:
# over 100 → "Large"
# 50 to 100 → "Medium"
# under 50 → "Small"

# Bonus: also add up the total of all orders and print it at the end.
order_label = "Unknown"
total_amount = 0.00

for o in orders:
    if o >= 100:
        order_label = "Large"
    elif o >= 50:
        order_label = "Medium"
    else:
        order_label = "Small"

    total_amount = total_amount + o
    print(f"${o} is {order_label}.")
print(f"Total amount is: ${total_amount:.2f}.")
print(f"Total amount is: ${sum(orders):.2f}.") # alternative way to calc total_amount
```
