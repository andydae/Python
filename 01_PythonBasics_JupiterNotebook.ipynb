```python
# print Hello World
print("Hello, world!");

# declare variables (don’t need declared types)
name = "Ava"           # str (like VARCHAR)
age = 34               # int
salary = 85000.50      # float (like DECIMAL)
is_active = True       # bool (like BIT);

print(name, age);
print(type(salary))    # <class 'float'>;

# f-strings are the cleanest way to build text
print(f"{name} is {age} years old");

# basic math
total = 10 + 5 * 2     # 20
avg = 15 / 4           # 3.75 (always a float)
whole = 15 // 4        # 3 (integer division)
remainder = 15 % 4     # 3 (modulo)

print(total, avg, whole, remainder);


# Exercise: create variables for a product name, price, and quantity.
# Then print a line like 3 x Widget = $29.97.

product_name = "Fishing Lure"
price = 12.99
quantity = 3

# Calculate total first before printing
total = quantity * price

print(f"{quantity} x {product_name} = ${total:.2f}")
```

```python
"""
IF, ELSEIF, ELSE: Python's version of CASE WHEN sql statements
"""

score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

print(grade)  # B

# Indentation matters here. The indented lines (4 spaces) are what belongs to each branch.
# There’s no BEGIN/END or {}. The colon : at the end of the line starts the block.
```

### Comparison and Logic Operators (Python vs SQL)

| Python | SQL | Meaning |
|---|---|---|
| `==` | `=` | Equal to |
| `!=` | `<>` or `!=` | Not equal to |
| `>` | `>` | Greater than |
| `<` | `<` | Less than |
| `>=` | `>=` | Greater than or equal to |
| `<=` | `<=` | Less than or equal to |
| `and` | `AND` | Both conditions true |
| `or` | `OR` | At least one condition true |
| `not` | `NOT` | Reverses a condition |
| `x in [1, 2, 3]` | `x IN (1, 2, 3)` | Value is in a list |
| `x not in [1, 2, 3]` | `x NOT IN (1, 2, 3)` | Value is not in a list |
| `x is None` | `x IS NULL` | Value is missing |
| `x is not None` | `x IS NOT NULL` | Value is present |

**Remember:** `=` assigns a value, `==` compares two values.
